```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/ecommerce_dataset.csv')
```

```python
df.drop(columns=['product_id','external_id','subcategory', 'url','images_count', 'first_image', 'tags' ,'scraped_at'], inplace=True)
```

```python
plt.figure(figsize=[6,6])

plt.hist(df['reviews_count'])
plt.title("Reviews Count", fontsize= 15)
plt.xlabel("Number of reviews", fontsize= 12)
plt.ylabel("Products count", fontsize= 12)

plt.axvline(np.mean(df['reviews_count']), color='r', linestyle = '--', label="Mean")
plt.axvline(np.median(df['reviews_count']), color='orange', linestyle = '--', label="Median")

plt.show()
```

```python
print(df["reviews_count"].mean())
print(df["reviews_count"].median())
print(df["reviews_count"].var())
print(df["reviews_count"].std())
```
