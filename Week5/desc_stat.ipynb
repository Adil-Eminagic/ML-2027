```python
import numpy as np
import matplotlib.pyplot as plt
```

### Mean

```python
def average(data):
    sum = 0

    for d in data:
        sum += d
    
    return sum / len(data)
```

```python
arr = np.random.random_integers(0, 10, 20)
print(arr)
```

```python
print(average(arr))
print(np.mean(arr))
```

```python
plt.figure(figsize=[6,6])

plt.hist(arr, bins=20)
plt.title("Histogram of random integers", fontsize= 15)
plt.xlabel("Position", fontsize= 12)
plt.ylabel("Count", fontsize= 12)

plt.show()
```

#### Weighted average

```python
weighted_arr = []

for i in range(11):
    count = 0

    for a in arr:
        if a == i:
            count += 1
    
    weighted_arr.append((i, count))
```

```python
weighted_arr
```

```python
wighted_sum = 0
total_weight = 0

for wa in weighted_arr:
    wighted_sum += wa[1] * wa[0]    # weight * position
    total_weight += wa[1]

print(f"Center of mass is {wighted_sum / total_weight}")
```

```python
# module 1 - 50 % exam, points 0-100
# module 2 - 35 % exam, points 0-100
# module 3 - 15 % exam, points 0-100

modul_weights = [50, 35, 15]
modul_points = [75, 25, 90]

total_points = [modul_weights[i] * modul_points[i]  for i in range(3)]

print(np.sum(total_points) / np.sum(modul_weights))
```

```python
# module 1 - 25 % exam, points 0-100
# module 2 - 25 % exam, points 0-100
# module 3 - 25 % exam, points 0-100
# module 4 - 25 % exam, points 0-100

modul_weights = np.full(4, 25)
modul_points = [75, 25, 90, 35]

total_points1 = [modul_weights[i] * modul_points[i]  for i in range(4)]
total_points2 = np.sum(modul_points) / 4

print(np.sum(total_points1) / np.sum(modul_weights))
print(total_points2)
```

#### What mean represent

```python
people = ['Tom', 'Diana', 'Harry', 'Sally']
heights = [1.75, 1.65, 1.80, 1.55]

plt.bar(people, heights, width=1)

plt.show()
```

```python
mean_height = average(heights)

plt.bar(people, np.full(len(people), mean_height), color='red', width=1)

plt.show()
```

```python
plt.bar(people, heights, width=1)

plt.bar(people, np.full(len(people), mean_height), color='red', width=1, alpha = 0.3)

plt.show()
```

### Median

```python
def median(sorted_arr):
    half_size = sorted_arr.size // 2 # 9 / 2 = 4, 8 / 2 = 4

    if half_size * 2 < sorted_arr.size:
        return sorted_arr[half_size]
    else:
        return (sorted_arr[half_size -1] + sorted_arr[half_size]) / 2
    
```

```python
print(median(np.sort(arr)))
print(np.median(np.sort(arr)))
```

```python
plt.figure(figsize=[6,6])

plt.hist(arr, bins=20)
plt.title("Histogram of random integers", fontsize= 15)

plt.axvline(average(arr), color='r', linestyle = '--', label="Mean")
plt.axvline(median(np.sort(arr)), color='orange', linestyle = '--', label="Median")

plt.xlabel("Position", fontsize= 12)
plt.ylabel("Count", fontsize= 12)

plt.legend()
plt.show()
```

### Variance and standard deviation

```python
def variance(data):

    mu = average(data)

    sum = 0

    for d in data:
        sum += (d - mu)**2

    return sum / len(data)

def standard_deviation(data):
    return np.sqrt(variance(data))
```

```python
print(variance(arr))
print(np.var(arr))
```

```python
print(standard_deviation(arr))
print(np.std(arr))
```
