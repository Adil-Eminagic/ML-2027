```python
import numpy as np
import matplotlib.pyplot as plt
```

## Fair dice

```python
fair_dice_probability = np.full(6, 1/6)

print(fair_dice_probability)
```

```python
plt.bar(range(1, 7), fair_dice_probability)
plt.xlabel('Dice Face')
plt.ylabel('Probability')
plt.title('Probability mass function pmf(x)')
plt.show()
```

```python
cummulative_value = 0

cumulative_probability =[]

for d in fair_dice_probability:
    cummulative_value += d
    cumulative_probability.append(cummulative_value)

cumulative_probability
```

```python
plt.bar(range(1, 7), cumulative_probability)
plt.xlabel('Fair Dice Face')
plt.ylabel('Cumulative distribution function')
plt.show()
```

## Unfair dice

```python
unfair_dice_probability = np.array([0.05, 0.15, 0.2, 0.1, 0.2, 0.3])
```

```python
plt.bar(range(1, 7), unfair_dice_probability)
plt.xlabel('Unfair Dice Face')
plt.ylabel('pmf(x) - Probability mass function')
plt.title('pmf - Probability mass function')
plt.show()
```

```python
cumulative_probability = np.cumsum(unfair_dice_probability)
print(cumulative_probability)
```

```python
plt.bar(range(1, 7), cumulative_probability)
plt.xlabel('Unfair Dice Face')
plt.ylabel('Cumulative Probability')
plt.show()
```
