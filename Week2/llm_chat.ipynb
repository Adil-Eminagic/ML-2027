```python
from openai import OpenAI
```

```python
model = "llama3.2:1b"

base_url = "http://localhost:11434/v1"

ollama = OpenAI(base_url=base_url, api_key="ollama")
```

```python
system_message = "You are helpful assistant"

system_mes = [{'role':'system', 'content': system_message}]
```

```python
messages = [
    {'role':'system', 'content': system_message},
    {'role':'user', 'content': "What is capital of France?"},
    {'role':'assistent', 'content': "The capital of France is Paris."},
    {'role':'user', 'content': "What was my last question?"}
]
```

```python
stream = ollama.chat.completions.create(
    model=model,
    messages=messages,
    stream= True
)
```

```python
full_text = ''

for chunk in stream:
    full_text += chunk.choices[0].delta.content or ''

print(full_text)
```
